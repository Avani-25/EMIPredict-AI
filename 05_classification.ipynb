# **Classification**

Loading the feature-engineered dataset

```python
import pandas as pd

df = pd.read_csv("feature_engineered_dataset.csv")

print("Dataset shape:", df.shape)
df.head()
```

Check the target

```python
print(df["emi_eligibility"].value_counts())
print(df["emi_eligibility"].value_counts(normalize=True))
```

Separate features and target

```python
X = df.drop("emi_eligibility", axis=1)
y = df["emi_eligibility"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("df shape:", df.shape)
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Recreate X and y from the CURRENT df

X = df.drop("emi_eligibility", axis=1)
y = df["emi_eligibility"]

print("df shape:", df.shape)
print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts(dropna=False))

print("\nTarget percentage:")
print(y.value_counts(normalize=True, dropna=False) * 100)
```

Split the data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Total rows in df:", len(df))
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
print("Missing target values:", y.isna().sum())
print("Total rows:", len(y))

print("\nTarget distribution:")
print(y.value_counts(dropna=False))
```

```python
print("Rows with missing target:")
print(df[df["emi_eligibility"].isna()].head())
```

# Import Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

print("Logistic Regression model created!")
```

```python
# Train it
logistic_model.fit(X_train, y_train)

print("Logistic Regression training completed!")
```

```python
# Make predictions
y_pred_logistic = logistic_model.predict(X_test)

print("Predictions completed!")
print("Number of predictions:", len(y_pred_logistic))
```

```python
# Calculate accuracy
from sklearn.metrics import accuracy_score

logistic_accuracy = accuracy_score(y_test, y_pred_logistic)

print("Logistic Regression Accuracy:", logistic_accuracy)
print("Accuracy (%):", logistic_accuracy * 100)
```

```python
# Full evaluation of Logistic Regression
from sklearn.metrics import classification_report, confusion_matrix

print("Classification Report:")
print(classification_report(y_test, y_pred_logistic))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_logistic))
```

**overall accuracy is 88.29%, but the model is poor at detecting class 1 That means Logistic Regression is detecting almost none of class 1. So 88.29% accuracy is misleading because class 2 is the dominant class.**

# Random Forest

```python
# Train Random Forest
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

print("Random Forest model created!")
```

```python
# train
rf_model.fit(X_train, y_train)

print("Random Forest training completed!")
```

```python
# predict
y_pred_rf = rf_model.predict(X_test)

print("Random Forest predictions completed!")
```

```python
# Calculate accuracy
from sklearn.metrics import accuracy_score

rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("Random Forest Accuracy:", rf_accuracy)
print("Accuracy (%):", rf_accuracy * 100)
```

```python
# Full evaluation
print("Random Forest Classification Report:")
print(classification_report(y_test, y_pred_rf))

print("\nRandom Forest Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
```

**Random Forest is much better overall than Logistic Regression.However, there is still a major issue: Class 1 (High_Risk) recall is only 0.07.**

**That means Random Forest correctly identifies only about 7% of the actual High_Risk cases.
So we should not stop at 95.24% accuracy.**

**Let's train a class-weighted Random Forest so mistakes on minority classes receive more importance.**

# Balanced Random Forest

```python
# Train weighted Random Forest
rf_balanced = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Balanced Random Forest created!")
```

```python
# Train
rf_balanced.fit(X_train, y_train)

print("Balanced Random Forest training completed!")
```

```python
# Predict
y_pred_rf_balanced = rf_balanced.predict(X_test)

print("Predictions completed!")
```

```python
# Evaluate
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

rf_balanced_accuracy = accuracy_score(y_test, y_pred_rf_balanced)

print("Balanced Random Forest Accuracy:",
      rf_balanced_accuracy)

print("Accuracy (%):",
      rf_balanced_accuracy * 100)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf_balanced))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf_balanced))
```

**The accuracy stayed at 95.24%, but the important improvement is the minority class.**

# **XGBoost Classifier**

```python
# Import XGBoost
from xgboost import XGBClassifier
```

```python
# Train XGBoost classifier
xgb_classifier = XGBClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=6,
    random_state=42,
    n_jobs=-1,
    eval_metric="logloss"
)

xgb_classifier.fit(
    X_train,
    y_train
)

print("XGBoost Classifier training completed!")
```

```python
# Make Prediction
# XGBoost was trained using the original feature columns
X_test_encoded = X_test

y_pred_xgb = xgb_classifier.predict(X_test_encoded)

y_prob_xgb = xgb_classifier.predict_proba(X_test_encoded)[:, 1]

print("Predictions completed!")
```

```python
# Evaluate
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

accuracy_xgb = accuracy_score(y_test, y_pred_xgb)

precision_xgb = precision_score(
    y_test,
    y_pred_xgb,
    average="weighted"
)

recall_xgb = recall_score(
    y_test,
    y_pred_xgb,
    average="weighted"
)

f1_xgb = f1_score(
    y_test,
    y_pred_xgb,
    average="weighted"
)

# Multiclass ROC-AUC
y_prob_xgb_multiclass = xgb_classifier.predict_proba(X_test_encoded)

roc_auc_xgb = roc_auc_score(
    y_test,
    y_prob_xgb_multiclass,
    multi_class="ovr",
    average="weighted"
)

print("XGBoost Classifier Results")
print("==========================")
print("Accuracy :", accuracy_xgb)
print("Precision:", precision_xgb)
print("Recall   :", recall_xgb)
print("F1 Score :", f1_xgb)
print("ROC-AUC  :", roc_auc_xgb)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_xgb))
```

# Create the final model comparison

```python
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Function to calculate multiclass metrics
def calculate_metrics(y_true, y_pred, model_name):
    
    accuracy = accuracy_score(y_true, y_pred)
    
    precision = precision_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )
    
    recall = recall_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )
    
    f1 = f1_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )
    
    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1
    }


# Calculate metrics for all available models
results = []

results.append(
    calculate_metrics(
        y_test,
        y_pred_logistic,
        "Logistic Regression"
    )
)

results.append(
    calculate_metrics(
        y_test,
        y_pred_rf,
        "Random Forest Classifier"
    )
)

results.append(
    calculate_metrics(
        y_test,
        y_pred_xgb,
        "XGBoost Classifier"
    )
)

results.append(
    calculate_metrics(
        y_test,
        y_pred_hgb,
        "HistGradientBoosting"
    )
)


classification_results = pd.DataFrame(results)

classification_results = classification_results.sort_values(
    by="F1_Score",
    ascending=False
).reset_index(drop=True)

display(classification_results)
```
