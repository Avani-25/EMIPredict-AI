```python
# Coverting the data into Numeric Columns
numeric_columns = [
    "age",
    "monthly_salary",
    "credit_score",
    "requested_amount",
    "requested_tenure"
]

for col in numeric_columns:
    df_clean[col] = (
        df_clean[col]
        .astype(str)
        .str.replace(r"\.0\.0$", "", regex=True)
    )
    
    df_clean[col] = pd.to_numeric(
        df_clean[col],
        errors="coerce"
    )

print("All 5 columns converted to numeric successfully!")

print("\nData types:")
print(df_clean[numeric_columns].dtypes)
```

```python
print(df_clean[numeric_columns].head())
```

```python
# Check whether values fall outside the ranges or not
print("Age range:")
print(df_clean["age"].astype(float).min(), 
      df_clean["age"].astype(float).max())

print("\nMonthly salary range:")
print(df_clean["monthly_salary"].astype(float).min(),
      df_clean["monthly_salary"].astype(float).max())

print("\nCredit score range:")
print(df_clean["credit_score"].astype(float).min(),
      df_clean["credit_score"].astype(float).max())

print("\nRequested amount range:")
print(df_clean["requested_amount"].astype(float).min(),
      df_clean["requested_amount"].astype(float).max())

print("\nRequested tenure range:")
print(df_clean["requested_tenure"].astype(float).min(),
      df_clean["requested_tenure"].astype(float).max())
```

```python
# ==========================================
# OUTLIER ANALYSIS USING IQR METHOD
# ==========================================
# IQR Method for Outlier Detection(Interquartile Range Method)

numerical_columns = [
    "age",
    "monthly_salary",
    "credit_score",
    "requested_amount",
    "requested_tenure"
]
for col in numerical_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    outliers = df_clean[(df_clean[col].astype(float) < lower_bound) | (df_clean[col].astype(float) > upper_bound)]
    print(f"\n---{col}---")
    print("Q1;",Q1)
    print("Q3:",Q3)
    print("IQR:",IQR)
    print("Lower Bound:",lower_bound)
    print("Upper Bound:",upper_bound)
    print("Number of outliers:", len(outliers))
```

```python
# ==========================================
# OUTLIER VALUE INSPECTION
# ==========================================

for col in numerical_columns:
    print(f"\n{'='*50}")
    print(f"{col.upper()}")
    print(f"{'='*50}")

    print("Minimum:", df_clean[col].astype(float).min())
    print("Maximum:", df_clean[col].astype(float).max())

    print("\nSelected percentiles:")
    print(df_clean[col].astype(float).quantile([0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]))
```

```python
# ==========================================
# CHECK SUSPICIOUS NUMERICAL VALUES
# ==========================================

# This will give us actual invalid-value counts, instead of blindly 
# deleting thousands of IQR outliers.

print("Credit score = 0:")
print((df_clean["credit_score"].astype(float) == 0).sum())

print("\nCredit score above 900:")
print((df_clean["credit_score"].astype(float) > 900).sum())

print("\nMonthly salary below 10000:")
print((df_clean["monthly_salary"].astype(float) < 10000).sum())

print("\nRequested amount below 10000:")
print((df_clean["requested_amount"].astype(float) < 10000).sum())

print("\nRequested tenure below 3:")
print((df_clean["requested_tenure"].astype(float) < 3).sum())

print("\nRequested tenure above 84:")
print((df_clean["requested_tenure"].astype(float) > 84).sum())
```

```python
# ==========================================
# CHECK CREDIT SCORE EXTREME VALUES
# ==========================================

print("Credit scores below 300:")
print((df_clean["credit_score"].astype(float) < 300).sum())

print("\nCredit scores from 300 to 850:")
print(
    (
        (df_clean["credit_score"].astype(float) >= 300) &
        (df_clean["credit_score"].astype(float) <= 850)
    ).sum()
)

print("\nCredit scores above 850:")
print((df_clean["credit_score"].astype(float) > 850).sum())

print("\nUnique credit scores above 850:")
print(
    sorted(
        df_clean.loc[
            df_clean["credit_score"].astype(float) > 850,
            "credit_score"
        ].astype(float).unique()
    )[:30]
)

# 1566 credit scores below 300 → invalid
# 397,604 between 300–850 → valid
# 3,210 above 850 → invalid
# The only values above 850 are 999 and 1200

# So we can safely treat 300–850 as the valid credit-score range for this cleaning step.
```

```python
# ==========================================
# FINAL NUMERICAL CLEANING
# ==========================================

# Convert numerical columns to numeric
numerical_columns = [
    "age",
    "monthly_salary",
    "credit_score",
    "requested_amount",
    "requested_tenure"
]

for col in numerical_columns:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")


# Remove invalid credit scores
df_clean = df_clean[
    (df_clean["credit_score"] >= 300) &
    (df_clean["credit_score"] <= 850)
].copy()


print("Invalid credit scores removed.")
print("Remaining rows:", len(df_clean))
```

```python
print("\n--- Final Numerical Validation ---")

print("Age:", df_clean["age"].min(), "to", df_clean["age"].max())

print(
    "Monthly Salary:",
    df_clean["monthly_salary"].min(),
    "to",
    df_clean["monthly_salary"].max()
)

print(
    "Credit Score:",
    df_clean["credit_score"].min(),
    "to",
    df_clean["credit_score"].max()
)

print(
    "Requested Amount:",
    df_clean["requested_amount"].min(),
    "to",
    df_clean["requested_amount"].max()
)

print(
    "Requested Tenure:",
    df_clean["requested_tenure"].min(),
    "to",
    df_clean["requested_tenure"].max()
)
```

```python
# ==========================================
# DUPLICATE CHECK
# ==========================================

duplicate_count = df_clean.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# ==========================================
# FINAL CLEANING
# ==========================================

# 1. Remove duplicate rows
df_clean = df_clean.drop_duplicates().copy()

# 2. Make sure numerical columns have correct numeric types
numerical_columns = [
    "age",
    "monthly_salary",
    "years_of_employment",
    "monthly_rent",
    "family_size",
    "dependents",
    "school_fees",
    "college_fees",
    "travel_expenses",
    "groceries_utilities",
    "other_monthly_expenses",
    "current_emi_amount",
    "credit_score",
    "emergency_fund",
    "requested_amount",
    "requested_tenure",
    "max_monthly_emi"
]

for col in numerical_columns:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")

# 3. Remove rows created with missing numerical values
df_clean = df_clean.dropna().copy()

# 4. Reset index
df_clean.reset_index(drop=True, inplace=True)

print("FINAL CLEANING COMPLETED ✅")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
print("Missing values:", df_clean.isnull().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
# ==========================================
# SAVE CLEAN DATASET
# ==========================================

output_path = "emi_prediction_cleaned.csv"

df_clean.to_csv(output_path, index=False)

print("Clean dataset saved successfully! ✅")
print("File:", output_path)
```
